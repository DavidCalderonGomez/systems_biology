```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['mathtext.fontset'] = 'cm'  # Computer Modern, mismo look que LaTeX
```

```python

# Read the whole file and split on whitespace (handles tabs and newlines)
with open("../data/single_run.txt", "r") as f:
    values = f.read().split()

values = np.array(values, dtype=float)

# Unpack: each triplet is (time, mRNA, protein)
time = values[0::3]
mRNA = values[1::3]
protein = values[2::3]
```

```python
fig, ax = plt.subplots(2, 1, figsize=(8,6), sharex=True)

ax[0].step(time, mRNA, where='post', color='tab:blue')
ax[0].set_ylabel(r"mRNA count")

ax[1].step(time, protein, where='post', color='tab:orange')
ax[1].set_ylabel(r"Protein count")
ax[1].set_xlabel(r"Time $t$")

plt.tight_layout()
plt.savefig("../figures/single_run.pdf")
plt.show()
```
